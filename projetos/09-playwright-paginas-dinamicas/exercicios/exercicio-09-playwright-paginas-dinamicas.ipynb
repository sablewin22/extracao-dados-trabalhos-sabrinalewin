## Sobre rodar no terminal

A API síncrona do Playwright **não** roda dentro de célula Jupyter. Seu código de coleta vai em `exercicio_09_script.py`, ao lado deste notebook. Você executa com:

```bash
uv run exercicio_09_script.py
```

Este notebook só orienta o roteiro e registra suas respostas. **Não** use `subprocess` aqui para a coleta.

## Parte 2: Escrever o script

Crie o arquivo `exercicio_09_script.py` **nesta mesma pasta** (`exercicios/`).

O script deve, em ordem:

1. Abrir a URL do interesse com Playwright sync (`headless=False`, para ver o navegador).
2. Fechar cookies/onboarding se aparecerem.
3. Esperar `[data-testid="story-item"]`.
4. Clicar **2 vezes** em More stories (com `scroll_into_view_if_needed()` antes do clique).
5. Coletar título + URL únicas → salvar `dados/exercicio-lista.csv`.
6. Para cada URL: abrir a story, extrair summary, métricas do painel e `source_urls` → salvar `dados/exercicio-detalhes.csv`.

8. Imprimir no terminal quantas linhas salvou em cada CSV.
