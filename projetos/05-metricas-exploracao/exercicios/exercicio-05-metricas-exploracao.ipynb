## Parte 3 — Transformar: qualidade e métrica

Confira duplicatas, remova-as se existirem, e calcule a taxa de engajamento, sempre lembrando de tirar as linhas com denominador igual a zero antes de dividir.

````python
duplicadas = df.duplicated().sum()  # conta linhas idênticas a outra linha anterior
print(f"Linhas duplicadas: {duplicadas}")

df = df.drop_duplicates()  # remove as duplicatas, se existirem
print(f"Linhas depois de remover duplicatas: {len(df)}")
````

````python
# complete aqui: filtre df pra manter só as linhas com "plays" maior que zero (ajuste o nome da coluna se a sua for diferente)
# guarde o resultado numa variável nova, por exemplo df_com_visualizacoes = df[...].copy()
df_com_visualizacoes = df[df["plays"] > 0].copy()
print(f"Linhas descartadas por plays = 0: {len(df) - len(df_com_visualizacoes)}")
````

````python
# complete aqui: crie a coluna "taxa_engajamento" em df_com_visualizacoes
# fórmula: (likes + comments + shares) / plays (ajuste os nomes de coluna se precisar)
# depois de completar, descomente a linha abaixo pra conferir o resultado:
# df_com_visualizacoes[["author", "likes", "plays", "taxa_engajamento"]].sort_values("taxa_engajamento", ascending=False).head(5)

df_com_visualizacoes["taxa_engajamento"] = (
    df_com_visualizacoes["likes"] + df_com_visualizacoes["comments"] + df_com_visualizacoes["shares"]
) / df_com_visualizacoes["plays"]  # a divisão inteira acontece linha a linha, automaticamente, sem precisar de laço

df_com_visualizacoes[["author", "likes", "plays", "taxa_engajamento"]].sort_values("taxa_engajamento", ascending=False).head(5)
````

## Parte 4 — Transformar: estatística descritiva

Calcule estatística descritiva das métricas de interação disponíveis na sua coleta.

````python
df_com_visualizacoes[["likes", "comments", "shares", "plays"]].describe()  # ajuste os nomes de coluna se a sua plataforma usar outros
````

````python
# complete aqui: calcule a média e a mediana de "likes" (ou equivalente) e imprima os dois valores
# compare os dois números: eles são parecidos ou bem diferentes? o que isso sugere sobre a distribuição?
# resposta: sugere que existem vídeos que possuem um nível de curtidas muito elevado comparado ao resto, os quais aumentam o valor da média e, consequentemente, colocam esse valor distante da mediana.

media = df_com_visualizacoes["likes"].mean()  
mediana = df_com_visualizacoes["likes"].median()  

print(f"Média de curtidas: {media:.1f}")
print(f"Mediana de curtidas: {mediana:.1f}")
````

## Parte 5 — Transformar: agrupamento por hashtag

Repita o processo da aula: separe a coluna de hashtags numa lista por linha, explode, e agrupe.

````python
df_hashtags = df_com_visualizacoes.dropna(subset=["hashtags"]).copy()  # mantém só linhas com hashtag preenchida

df_hashtags["hashtags"] = df_hashtags["hashtags"].str.split(",")  # separa a string em lista, dividindo pela vírgula

df_hashtags = df_hashtags.explode("hashtags")  # transforma cada item da lista numa linha própria

df_hashtags["hashtags"] = df_hashtags["hashtags"].str.strip()  # remove espaço em volta da hashtag

print(f"Linhas depois de explodir: {len(df_hashtags)}")
df_hashtags[["author", "hashtags", "likes"]].head(10)
````

````python
# complete aqui: agrupe df_hashtags por "hashtags" com .groupby().agg(), calculando:
# qtd_posts (contagem, use a coluna "id" ou outra que sirva de identificador único)
# engajamento_medio (média de "taxa_engajamento")
# curtidas_medias (média de "likes")
# use .reset_index() no final, como na aula, e guarde o resultado em resumo_hashtags

resumo_hashtags = df_hashtags.groupby("hashtags").agg(  # agrupa as linhas pela coluna "hashtags"
    qtd_posts=("id", "count"),  # conta quantos posts (linhas) cada hashtag teve
    curtidas_medias=("likes", "mean"),  # média de curtidas dos posts daquela hashtag
    engajamento_medio=("taxa_engajamento", "mean"),  # média da taxa de engajamento dos posts daquela hashtag
).reset_index()  # reset_index() transforma a hashtag (que virou índice do agrupamento) de volta numa coluna normal

print(f"Hashtags distintas na coleta: {len(resumo_hashtags)}")
resumo_hashtags.head(15)  # amostra; a tabela completa entra na ordenação da próxima seção
````

## Parte 6 — Conferir e responder a pergunta

Volte pra pergunta que você definiu na Parte 0. Use `resumo_hashtags` (ordenando com `.sort_values()`, filtrando ou comparando linhas) pra respondê-la com código.

````python
# escreva aqui o código que responde à sua pergunta, usando resumo_hashtags
# Vídeos com hashtags relacionadas à política tiveram mais curtidas ou vídeos com hashtags relacionadas a entretenimento tiveram mais curtidas?

tabela_final = resumo_hashtags.sort_values("curtidas_medias", ascending=False)  # ordena da hashtag com maior engajamento médio pra menor

tabela_final = tabela_final.rename(columns={  # troca os nomes técnicos das colunas por nomes mais claros pra um relatório
    "hashtags": "Hashtag",
    "qtd_posts": "Qtd. de posts",
    "curtidas_medias": "Curtidas médias",
    "engajamento_medio": "Engajamento médio",
    "plays_total": "Visualizações (total)",
})

tabela_final["Curtidas médias"] = (tabela_final["Curtidas médias"] * 100).round(1)  # converte a taxa (0.08) pra porcentagem (8.0), mais fácil de ler

print(f"Linhas na tabela-resumo: {len(tabela_final)}")
tabela_final.head(15)  
````

**Interpretação, em texto:**

> A tabela gerada mostra que vídeos com hashtags relacionadas a entretenimento possuem mais curtidas do que vídeos com hashtags de política (responde a pergunta criada). Além disso, ela mostra que a hashtag "edit" é a que gera mais curtidas, porém não é a que possui a maior taxa de engajamento, provando que o engajamento não se baseia somente em curtidas. A tabela não permite concluir que as hashtags possuem alguma relevância nas visualizações totais.



## Parte 7 — Salvar e comunicar

Monte a tabela-resumo final (ordenada, com nomes de coluna claros) e salve como CSV, do mesmo jeito que na aula.

````python
tabela_final = resumo_hashtags.sort_values("engajamento_medio", ascending=False)  # ordena da hashtag com maior engajamento médio pra menor

tabela_final = tabela_final.rename(columns={  # ajuste os nomes conforme as colunas que você usou acima
    "hashtags": "Hashtag",
    "qtd_posts": "Qtd. de posts",
    "curtidas_medias": "Curtidas médias",
    "engajamento_medio": "Engajamento médio",
})

tabela_final
````

````python
tabela_final.to_csv("dados/resumo_hashtags.csv", index=False)  # salva a tabela final, sem a coluna de índice numérico
print("Tabela salva em dados/resumo_hashtags.csv")
````
